```python
import polars as pl
lf = pl.scan_parquet("../../data/output/relevant_reviews.parquet")

display(lf.select(pl.len()).collect())                     
display(lf.head(20).collect())                            

display(lf.filter(pl.col("relevance_score") >= 0.5)
          .sort("relevance_score", descending=True)
          .select("relevance_score", "review_text")
          .head(20).collect())

# ini yg potentially dibuang karena agak generik, bisa diperbaiki si lewat annotator
display(lf.filter(~pl.col("is_relevant"))
          .sort("relevance_score", descending=True)
          .select("relevance_score", "review_text")
          .head(20).collect())
```
